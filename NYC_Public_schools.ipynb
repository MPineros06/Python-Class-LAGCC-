## Comparison with Other NYC Schools

```python
df5 =df[(df["Regents Exam"] == "Common Core Algebra") & (df["School Name"] != "Academy of Finance and Enterprise")].copy()
```

```python
df5 =df5[df5["Mean Score"] != "s"].copy()
```

```python
df5["Mean Score"] = df5["Mean Score"].astype(float)
```

```python
df5.groupby("Year")["Mean Score"].mean()
```

```python
school_average = df4.groupby("Year")["Mean Score"].mean()
nyc_average = df5.groupby("Year")["Mean Score"].mean()

comparison = pd.DataFrame({
    "Academy of Finance and Enterprise": school_average,
    "Other NYC Schools": nyc_average
})
```

```python
import matplotlib.pyplot as plt
```

```python
comparison.plot(kind="bar", figsize=(10, 6), color=["steelblue", "gray"])
plt.title("Common Core Algebra Mean Score Comparison")
plt.ylabel("Mean Score")
plt.xlabel("Year")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(loc="upper right")
plt.show()
```

## Findings
Academy of finance and Enterprise had a higher average common core than the other NYC Schools un every year analyzed.
In 2015, the school's average was 68.80, compared with 63.16 for the other schools.

```python
borough_df = df [df["Regents Exam"] == "Common Core Algebra"].copy()
```

```python
borough_df = borough_df[borough_df["Mean Score"] != "s"].copy()
```

```python
borough_df["Mean Score"] = borough_df["Mean Score"].astype(float)
```

```python
borough_df["Borough"] = borough_df["School DBN"].str[2].map({"M": "Manhattan", "X": "Bronx", "K": "Brooklyn", "Q": "Queens", "R": "Staten Island"})
```

```python
borough_average = borough_df.groupby("Borough")["Mean Score"].mean()
```

```python
borough_average
```

```python
borough_average.sort_values(ascending=False).plot(kind="bar", figsize=(10, 6), color="steelblue")
plt.title("Common Core Algebra Mean Score by Borough")
plt.ylabel("Mean Score")
plt.xlabel("Borough")
plt.ylim(0, 100)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## Findings
Staten Island had a higher average common core, while the Bronx had the lowest.
