```python
df.groupby('Gender')['salary'].median().sort_values(ascending=False).plot(kind='bar', color=['lightseagreen', 'steelblue']).set_title('Median Salarie by Gender')
```

Male professors have a slightly higher median salary than female professors.

```python
# Calculate the average salary by gender
# Creat a bar chart to compare the results
df.groupby('Gender')['salary'].mean().sort_values(ascending=True).plot(kind='bar', color=['lightseagreen', 'steelblue']).set_title('Average Salary by Gender')
```

```python
#Calculate the average salary by rank
#Create a bar chart to compare the results
df.groupby('rank')['salary'].mean().sort_values(ascending=False).plot(kind='bar', color=['lightseagreen', 'steelblue', 'slateblue']).set_title('Average Salary by Rank')
```

Professors have the highest average salary, followed by Associate Professors.
Assistant Professors hav the lowest average salary.
